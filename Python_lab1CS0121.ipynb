Show how to use a iterator. First, use enumerate to create a new tuple from above list

```python
# use of an iterator
# Iterator returns numbers that starts with 25
# and each time it is passed, it gets squared.

class MyInit:
    def __init__(self):
        self.a = 25

    def __iter__(self):
        return self

    def __next__(self):
        x = self.a
        self.a = self.a ** 2
        return x


obj1 = MyInit()
itr = iter(obj1)

print(next(itr))
print(next(itr))
print(next(itr))
```

Find sum of each line, mean, median and standard deviation of all lines in the file.

```python
import math

line_lengths = []

with open('poem.txt', 'r') as file:
    for line in file:
        length = len(line.rstrip('\n'))
        line_lengths.append(length)

print("Line lengths:", line_lengths)
print("Sum:", sum(line_lengths))

mean = sum(line_lengths) / len(line_lengths)
print("Mean:", mean)

sorted_lengths = sorted(line_lengths)
n = len(sorted_lengths)

if n % 2 == 0:
    median = (sorted_lengths[n // 2 - 1] + sorted_lengths[n // 2]) / 2
else:
    median = sorted_lengths[n // 2]

print("Median:", median)

variance = sum((x - mean) ** 2 for x in line_lengths) / len(line_lengths)
standard_deviation = math.sqrt(variance)

print("Standard Deviation:", standard_deviation)
```

# Working with CSV

Let's import a file with fuel economy data

```python
import csv

with open('mpg.csv') as csvfile:
    mpg = list(csv.DictReader(csvfile))
```

Find out how many items are there in the dictionary

```python
for item in mpg[1].items():
    print(item)
```

Find the column names with keys

```python
print(mpg[0].keys())
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```python
total = 0

for d in mpg:
    total += float(d['cty'])

average_cty = total / len(mpg)

print("Average cty fuel economy:", average_cty)
```

Find the average hwy fuel economy across all cars

```python
total = 0

for d in mpg:
    total += float(d['hwy'])

average_hwy = total / len(mpg)

print("Average hwy fuel economy:", average_hwy)
```

Use set to return the unique values for the number of cylinders the cars.

```python
cylinders = set(d['cyl'] for d in mpg)

print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```python
MpgByCyl = []

for c in cylinders:
    summpg = 0
    typecount = 0

    for d in mpg:
        if d['cyl'] == c:
            summpg += float(d['cty'])
            typecount += 1

    MpgByCyl.append((c, summpg / typecount))

MpgByCyl.sort(key=lambda x: x[0])

MpgByCyl
```

Return the unique values for the class types in our dataset using set.

```python
vehicleclass = set(d['class'] for d in mpg)

print(vehicleclass)
```

Find average miles per gallon in the highway for each class of vehicles

```python
MileageByClass = []

for t in vehicleclass:
    sumpg = 0
    classCount = 0

    for d in mpg:
        if d['class'] == t:
            sumpg += float(d['hwy'])
            classCount += 1

    MileageByClass.append((t, sumpg / classCount))

MileageByClass.sort(key=lambda x: x[1])

MileageByClass

        
```
